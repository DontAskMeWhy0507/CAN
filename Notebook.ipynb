```python
import platform
import psutil
import os

print("="*20, " SYSTEM INFORMATION ", "="*20)
# Operating System
print(f"OS: {platform.system()} {platform.release()} ({platform.machine()})")
print(f"Kernel Version: {platform.version()}")

# CPU Details
print("\n" + "="*20, " CPU INFO ", "="*20)
# Get CPU model name from linux system
cpu_model = os.popen("cat /proc/cpuinfo | grep 'model name' | uniq").read().strip().split(": ")
print(f"Model: {cpu_model[1] if len(cpu_model) > 1 else 'Unknown'}")
print(f"Physical Cores: {psutil.cpu_count(logical=False)}")
print(f"Total (Logical) Cores: {psutil.cpu_count(logical=True)}")

# RAM Details
print("\n" + "="*20, " MEMORY INFO ", "="*20)
svmem = psutil.virtual_memory()
print(f"Total RAM: {svmem.total / (1024**3):.2f} GB")
print(f"Available RAM: {svmem.available / (1024**3):.2f} GB")
print(f"Used RAM: {svmem.used / (1024**3):.2f} GB ({svmem.percent}%)")

# Disk Space
print("\n" + "="*20, " DISK INFO ", "="*20)
disk_usage = psutil.disk_usage('/')
print(f"Total Space: {disk_usage.total / (1024**3):.2f} GB")
print(f"Used Space: {disk_usage.used / (1024**3):.2f} GB ({disk_usage.percent}%)")
print(f"Free Space: {disk_usage.free / (1024**3):.2f} GB")
```
