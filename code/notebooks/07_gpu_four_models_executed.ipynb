Tái sử dụng các cell GPU đã hoàn tất với manifest gốc nguyên vẹn. Chỉ cho khác danh sách mô hình được lập lịch; code, data, môi trường, seed và hyperparameters phải khớp. AssocRules chạy trong notebook 09 song song.

```python
from pathlib import Path
import subprocess,sys,json
CODE=Path(r'F:/00_Research/code')
OUTPUT=CODE/'results/comparison_gpu4_reuse_v1'
REUSE=CODE/'results/comparison_content_v1_5models_backup'
import torch
assert torch.cuda.is_available(), 'Không chạy full bằng CPU ngoài ý muốn'
print('Python:',sys.executable)
print('GPU:',torch.cuda.get_device_name(0))
print('Output:',OUTPUT)
print('Giữ lại các cell đã hoàn tất:',REUSE)
```

```python
subprocess.run([sys.executable,str(CODE/'scripts/run_gpu_comparisons_reuse.py'),'--stage','run','--reuse-from',str(REUSE),'--output',str(OUTPUT)],check=True)
```

```python
items=[json.loads(p.read_text()) for p in OUTPUT.glob('set_*/seed_*/*/attempt_*/manifest.json')]
complete=[m for m in items if m['status']=='COMPLETE']
cells={(m['signature']['set'],m['signature']['seed'],m['method']) for m in complete}
assert len(cells)==48, f'Còn thiếu cell: {len(cells)}/48'
print('Đủ 48 cell GPU; bốn cell đầu được giữ lại, không huấn luyện lại.')
```
